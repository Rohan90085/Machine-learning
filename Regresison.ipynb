````python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score  # added r2_score

d = {
    'hour': [2, 3, 4, 5, 6, 7],
    'atndc': [50, 90, 78, 43, 56, 89],
    'marks': [70, 98, 85, 45, 59, 90]
}
df = pd.DataFrame(d)

# 1. Split data
xt, xtst, yt, ytst = train_test_split(df[['hour','atndc']], df['marks'], test_size=0.2, random_state=42)

# 2. Train
model = LinearRegression()
model.fit(xt, yt)

# 3. Predict
y_pred = model.predict(xtst)

# 4. Metrics
mse = mean_squared_error(ytst, y_pred)
r2 = r2_score(ytst, y_pred)

print("Mean Squared Error:", mse)
print("R^2 Score:", r2)#if it is equall to ~1 then it is good model

# 5. Manual prediction
time = int(input("enter the hours studied>"))
atendc = int(input("enter the attendance>"))
print("\nmodel prediction:", model.predict([[time, atendc]]))

print('\nslope:', model.coef_)
print('intercept:', model.intercept_)
````

````python
# # Linear Regression

# ## What is Linear Regression?

# Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value based on one or more input features. It finds the best-fit straight line that represents the relationship between the independent variables (features) and the dependent variable (target).

# ### Real-World Example

# Suppose we want to predict a student's marks based on the number of hours studied.

# | Hours Studied | Marks |
# | ------------- | ----- |
# | 1             | 35    |
# | 2             | 45    |
# | 3             | 55    |
# | 4             | 65    |
# | 5             | 75    |

# Linear Regression learns the pattern and creates a line that can predict marks for new study hours.

# ## Equation of Linear Regression

# y = mx + b

# Where:

# * **y** = Predicted value (output)
# * **x** = Input feature
# * **m** = Slope of the line
# * **b** = Intercept (value of y when x = 0)

# ## How It Works

# 1. Collect training data.
# 2. Find the best-fit line through the data points.
# 3. Minimize the prediction error (residuals).
# 4. Use the learned line to predict future values.

# ## Important Terms

# ### Feature (X)

# Input variable used for prediction.
# Example: Hours Studied

# ### Target (Y)

# Output variable to be predicted.
# Example: Marks

# ### Prediction

# The value predicted by the model.

# ### Residual (Error)

# Difference between actual value and predicted value.

# Residual = Actual Value - Predicted Value

# ### Best-Fit Line

# The line that minimizes the sum of squared errors.

# ## Types of Linear Regression

# ### 1. Simple Linear Regression

# Uses one input feature.

# Example:

# * Hours Studied → Marks

# ### 2. Multiple Linear Regression

# Uses multiple input features.

# Example:

# * Hours Studied
# * Sleep Hours
# * Attendance

# to predict Marks.

# Equation:

# y = b0 + b1x1 + b2x2 + ... + bnxn

# ## Advantages

# * Easy to understand and implement.
# * Fast training process.
# * Works well for linear relationships.
# * Interpretable results.

# ## Limitations

# * Assumes a linear relationship.
# * Sensitive to outliers.
# * Performance decreases for complex nonlinear data.

# ## Scikit-Learn Implementation

# ```python
# from sklearn.linear_model import LinearRegression

# X = df[["Hours"]]
# y = df["Marks"]

# model = LinearRegression()
# model.fit(X, y)

# prediction = model.predict([[6]])
# print(prediction)
# ```

# ## Evaluation Metrics

# ### Mean Absolute Error (MAE)

# Average absolute prediction error.

# ### Mean Squared Error (MSE)

# Average squared prediction error.

# ### R² Score

# Measures how well the model explains the variance in the data.

# * R² = 1 → Perfect prediction
# * R² = 0 → No relationship

# ## Conclusion

# Linear Regression is one of the most fundamental machine learning algorithms. It is widely used for predicting continuous values such as house prices, salaries, sales, temperatures, and student marks. It serves as the foundation for understanding more advanced regression techniques in machine learning.
````
