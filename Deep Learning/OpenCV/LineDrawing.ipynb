```python
import cv2
```

```python
#cv2.line(img,start,end,color,thck)

img=cv2.imread("Photos\Ganesha.png")
cv2.imshow("ganesha",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
print(img.shape)
```

```python
linedimg=cv2.line(img,(200,200),(200,200),(255,0,0),3)
cv2.imshow("l",linedimg)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rectangle=cv2.rectangle(img,(50,50),(200,200),(0,255,0),3)
cv2.imshow("rectangle",rectangle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
