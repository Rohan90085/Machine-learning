```python
import cv2

img = cv2.imread(
    r"C:\Users\Abhilash PH\OneDrive - REVA University\Desktop\scholar\12TH MarksCrad.jpeg",
    0
)
#Contrast Limited Adaptive Histogram Equalization.
clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8,8)
)

result = clahe.apply(img)

cv2.imwrite("clahe_output.jpg", result)

print("Saved Successfully")
```
