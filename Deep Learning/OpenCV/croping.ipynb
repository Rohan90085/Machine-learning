```python
import cv2
```

```python
img=cv2.imread("C:\\Users\\Abhilash PH\\OneDrive - REVA University\\Desktop\\Machine-learning\\Deep Learning\\OpenCV\\Photos\\Ganesha.png")
cv2.imshow("Ganesha",img)
print("Size:",img.shape)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
#cv2.resize(src,dimensionSize)
```

```python
resize=cv2.resize(img,(25,25))
cv2.imshow("resized",resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imwrite("C:\\Users\\Abhilash PH\\OneDrive - REVA University\\Desktop\\Machine-learning\\Deep Learning\\OpenCV\\Photos\\Ganesha_resized.png",resize)
print("after:",resize.shape)
```

```python
#Cropping
#cropped = img[y1:y2, x1:x2]
print("before:",img.shape)
crop=img[200:400,200:400]
cv2.imshow("cropped",crop)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imwrite("C:\\Users\\Abhilash PH\\OneDrive - REVA University\\Desktop\\Machine-learning\\Deep Learning\\OpenCV\\Photos\\croped.jpg",crop)

print("after:",crop.shape)
```

```python
#roatating
vertical_flip=cv2.flip(img,0)
cv2.imshow("vertical_flip",vertical_flip)
horizontal_flip=cv2.flip(vertical_flip,1)
cv2.imshow("horizontal_flip",horizontal_flip)
flip_both=cv2.flip(img,-1)#mirror image right to left
cv2.imshow("filppeed both",flip_both)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
