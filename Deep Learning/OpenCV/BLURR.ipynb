```python
import cv2
img=cv2.imread("Photos\Ganesha.png")
cv2.imshow("image",img)
#cv2.GaussianBlur(img,kernel_size=(11,11),sigmaX=0) kernelsize directly proportinal to blur
blur=cv2.GaussianBlur(img,(11,11),-1)
cv2.imshow("blurred",blur)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#thresholding
import cv2
img=cv2.imread("Photos\Ganesha.png")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
#cv2.threshold(img,threshold,color_intensity,cv2.THRESH_BINARY) if pixel>threshold then colorwhite,pixel<threshold then colorblack
thr=cv2.threshold(gray,60,160,cv2.THRESH_BINARY)[1]
cv2.imshow("thresholded",thr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Adaptive thresholding
#*best for shadow and the uneven ligth
#automatically calucaltes the thresh by comparing the neighbour pixel values and then applies the thresholding
import cv2

img=cv2.imread("Photos\Ganesha.png")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2.imshow("gray",gray)
#cv2.adaptivethreshold(img,pixel_Color,automatcaly caluculates the neighbour thrsh value))
thresh=cv2.adaptiveThreshold(gray,200,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,11,3)
cv2.imshow("adaptive threshold",thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img = cv2.imread("Photos/Ganesha.png")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2.imshow("gray",gray)

ret, thresh = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)
cv2.imshow("Otsu's thresholding", thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
