```python
#marks predicting through the deep learning model
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from tensorflow.keras import layers
import numpy as np
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv('C:\\Users\\Abhilash PH\\OneDrive - REVA University\\Desktop\\Machine-learning\\projects\\Final_Marks_Data.csv')
print(df)
```

```python
df.columns

X=df[['Attendance (%)', 'Internal Test 1 (out of 40)',
       'Internal Test 2 (out of 40)', 'Assignment Score (out of 10)',
       'Daily Study Hours']]
Y=df['Final Exam Marks (out of 100)']
```

```python
xtr, xte, ytr, yte = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
model=keras.Sequential([
    layers.Input(shape=(xtr.shape[1],)),
    layers.Dense(64,activation='relu'),
    layers.Dense(1)
])
```

```python
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
model.summary()
```

```python
#dont run this command until u check the capacity of your laptop
earlyStopping=EarlyStopping(
    monitor='val_Loss',
    patience=10,
    restore_best_weights=True
)
history=model.fit(
    xtr,ytr,epochs=450,batch_size=32,validation_split=0.2,callbacks=[earlyStopping]
)
```

```python
print(xtr.shape)
print(xte.shape)
print(ytr.shape)
print(yte.shape)
```

```python
p=model.predict(xte).flatten()
print(p[0])
print(r2_score(yte,p))
```
