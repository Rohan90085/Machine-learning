````python
# # Random Forest

# ## Introduction

# Random Forest is a supervised machine learning algorithm used for both **classification** and **regression** problems. It is an ensemble learning technique that combines multiple Decision Trees and produces a final prediction based on voting (classification) or averaging (regression).

# Random Forest was developed to overcome the problem of overfitting in a single Decision Tree.

# ---

# ## Why Random Forest?

# A Decision Tree can easily memorize training data and may perform poorly on unseen data. Random Forest solves this by creating many Decision Trees and combining their predictions.

# ### Advantages over Decision Tree

# * Reduces overfitting
# * Higher accuracy
# * Better generalization
# * Handles large datasets efficiently
# * Works with missing values and noisy data

# ---

# ## How Random Forest Works

# ### Step 1: Bootstrap Sampling

# Random Forest creates multiple datasets by randomly selecting records from the original dataset with replacement.

# Example:

# Original Data:

# | Student | Marks |
# | ------- | ----- |
# | A       | 80    |
# | B       | 70    |
# | C       | 90    |
# | D       | 60    |

# Tree 1 Sample:

# | Student | Marks |
# | ------- | ----- |
# | A       | 80    |
# | B       | 70    |
# | A       | 80    |
# | D       | 60    |

# Tree 2 Sample:

# | Student | Marks |
# | ------- | ----- |
# | C       | 90    |
# | D       | 60    |
# | B       | 70    |
# | C       | 90    |

# Each tree receives a different dataset.

# ---

# ### Step 2: Random Feature Selection

# Instead of considering all features at every split, Random Forest randomly selects a subset of features.

# Example:

# Features:

# * Hours Studied
# * Attendance
# * Internal Assessment
# * Sleep Hours
# * Assignments

# Tree 1 may use:

# * Hours Studied
# * Attendance

# Tree 2 may use:

# * Attendance
# * Sleep Hours

# Tree 3 may use:

# * Internal Assessment
# * Assignments

# This creates diversity among trees.

# ---

# ### Step 3: Build Multiple Decision Trees

# Each tree is trained independently using different data and different feature subsets.

# Example:

# Tree 1 → Pass

# Tree 2 → Fail

# Tree 3 → Pass

# Tree 4 → Pass

# Tree 5 → Fail

# ---

# ### Step 4: Final Prediction

# #### Classification

# Uses Majority Voting.

# Example:

# Pass = 3 votes

# Fail = 2 votes

# Final Prediction = Pass

# #### Regression

# Uses Average.

# Example:

# Tree Predictions:

# 70, 75, 80, 78, 77

# Average:

# (70 + 75 + 80 + 78 + 77) / 5 = 76

# Final Prediction = 76

# ---

# ## Important Parameters

# ### n_estimators

# Number of trees in the forest.

# ```python
# RandomForestClassifier(n_estimators=100)
# ```

# More trees generally improve performance but increase training time.

# ---

# ### max_depth

# Maximum depth of each tree.

# ```python
# RandomForestClassifier(max_depth=5)
# ```

# Controls overfitting.

# ---

# ### min_samples_split

# Minimum samples required to split a node.

# ```python
# RandomForestClassifier(min_samples_split=4)
# ```

# ---

# ### min_samples_leaf

# Minimum samples required in a leaf node.

# ```python
# RandomForestClassifier(min_samples_leaf=2)
# ```

# ---

# ### max_features

# Number of features considered at each split.

# ```python
# RandomForestClassifier(max_features='sqrt')
# ```

# Common values:

# * sqrt
# * log2
# * None

# ---

# ### random_state

# Ensures reproducible results.

# ```python
# RandomForestClassifier(random_state=42)
# ```

# ---

# ## Classification Example

# ### Dataset

# | Hours | Attendance | IA | Pass |
# | ----- | ---------- | -- | ---- |
# | 1     | 100        | 19 | 0    |
# | 2     | 150        | 20 | 0    |
# | 3     | 200        | 21 | 0    |
# | 4     | 250        | 22 | 1    |
# | 5     | 300        | 23 | 1    |
# | 6     | 350        | 24 | 1    |
# | 7     | 400        | 25 | 1    |

# ### Code

# ```python
# import pandas as pd
# from sklearn.ensemble import RandomForestClassifier

# d = {
#     'hour':[1,2,3,4,5,6,7],
#     'attendance':[100,150,200,250,300,350,400],
#     'ia':[19,20,21,22,23,24,25],
#     'pass':[0,0,0,1,1,1,1]
# }

# df = pd.DataFrame(d)

# X = df[['hour','attendance','ia']]
# y = df['pass']

# model = RandomForestClassifier(
#     n_estimators=100,
#     random_state=42
# )

# model.fit(X,y)

# prediction = model.predict([[4,240,22]])

# print(prediction)
# ```

# ---

# ## Feature Importance

# Random Forest can determine which features are most important.

# ```python
# print(model.feature_importances_)
# ```

# Example Output:

# ```python
# [0.65 0.25 0.10]
# ```

# Meaning:

# * Hours = 65%
# * Attendance = 25%
# * IA = 10%

# Hours contributes the most to prediction.

# ---

# ## Evaluation Metrics

# ### Accuracy

# ```python
# from sklearn.metrics import accuracy_score

# accuracy = accuracy_score(y_test, y_pred)
# ```

# Measures overall correctness.

# ---

# ### Precision

# ```python
# from sklearn.metrics import precision_score
# ```

# Measures how many predicted positives are actually positive.

# ---

# ### Recall

# ```python
# from sklearn.metrics import recall_score
# ```

# Measures how many actual positives are correctly identified.

# ---

# ### F1 Score

# ```python
# from sklearn.metrics import f1_score
# ```

# Balance between Precision and Recall.

# ---

# ### Confusion Matrix

# ```python
# from sklearn.metrics import confusion_matrix
# ```

# Shows:

# * True Positive
# * True Negative
# * False Positive
# * False Negative

# ---

# ## Advantages

# 1. High accuracy
# 2. Reduces overfitting
# 3. Handles large datasets
# 4. Works for classification and regression
# 5. Handles missing values
# 6. Provides feature importance
# 7. Robust against noise

# ---

# ## Disadvantages

# 1. Slower than a single Decision Tree
# 2. Requires more memory
# 3. Less interpretable
# 4. Training time increases with more trees

# ---

# ## Difference Between Decision Tree and Random Forest

# | Decision Tree     | Random Forest       |
# | ----------------- | ------------------- |
# | Single tree       | Multiple trees      |
# | High overfitting  | Reduced overfitting |
# | Faster            | Slower              |
# | Lower accuracy    | Higher accuracy     |
# | Easy to interpret | Harder to interpret |
# | Unstable          | More stable         |

# ---

# ## Real-World Applications

# * Loan Approval
# * Credit Risk Analysis
# * Fraud Detection
# * Medical Diagnosis
# * Customer Churn Prediction
# * Recommendation Systems
# * Stock Market Prediction
# * Employee Attrition Analysis

# ---

# ## Interview Questions

# ### What is Random Forest?

# Random Forest is an ensemble learning algorithm that combines multiple Decision Trees and uses majority voting or averaging for final predictions.

# ### Why is it called Random Forest?

# Because each tree is built using random samples of data and random subsets of features.

# ### What is Bootstrap Sampling?

# Random selection of training records with replacement.

# ### What is Majority Voting?

# The class receiving the highest number of votes from all trees becomes the final prediction.

# ### Why does Random Forest perform better than a Decision Tree?

# Because it reduces overfitting and combines knowledge from multiple trees, leading to better generalization.
````
