```python
import cv2
import numpy as np

img = cv2.imread(r"C:\Users\Abhilash PH\OneDrive - REVA University\Desktop\Machine-learning\Deep Learning\OpenCV\Photos\Ganesha.png")

pts1 = np.float32([
    [100,100],
    [400,100],
    [100,500],
    [400,500]
])

pts2 = np.float32([
    [0,0],
    [300,0],
    [0,400],
    [300,400]
])

matrix = cv2.getPerspectiveTransform(
    pts1,
    pts2
)

result = cv2.warpPerspective(
    img,
    matrix,
    (300,400)
)

cv2.imshow("Original", img)
cv2.imshow("Scanned", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

img = cv2.imread(r"C:\Users\Abhilash PH\OneDrive - REVA University\Desktop\Machine-learning\Deep Learning\OpenCV\Photos\Ganesha.png")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 100, 200)

cv2.imshow("Original", img)
cv2.imshow("Edges", edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
