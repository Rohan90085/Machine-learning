````python
import pandas as pd
from sklearn.linear_model import LogisticRegression
d={
    'marks':[35, 45, 55, 65, 75],
    'result':[0, 0, 0, 1, 1]
}
df=pd.DataFrame(d)
model=LogisticRegression()
model.fit(df[['marks']],df['result'])
mark=int(input("enter the marks:"))
p=model.predict([[mark]])[0]
if(p==1):
    print("pass")
else:   
    print("fail")
````

````python
# # Logistic Regression

# ## What is Logistic Regression?

# Logistic Regression is a supervised machine learning algorithm used for **classification problems**. Unlike Linear Regression, which predicts continuous values, Logistic Regression predicts the probability of a data point belonging to a particular class.

# ### Real-World Example

# Suppose we want to predict whether a student will pass or fail based on study hours.

# | Hours Studied | Result   |
# | ------------- | -------- |
# | 1             | Fail (0) |
# | 2             | Fail (0) |
# | 3             | Pass (1) |
# | 4             | Pass (1) |
# | 5             | Pass (1) |

# The output is categorical:

# * 0 = Fail
# * 1 = Pass

# Therefore, Logistic Regression is used instead of Linear Regression.

# ---

# ## Why Not Linear Regression?

# Linear Regression can predict values less than 0 or greater than 1, which does not make sense for probabilities.

# Example:

# * Probability = 1.5 ❌
# * Probability = -0.3 ❌

# Logistic Regression solves this problem by using the **Sigmoid Function**, which always gives outputs between 0 and 1.

# ---

# ## Sigmoid Function

# The sigmoid function converts any value into a probability between 0 and 1.

# [
# P(y=1)=\frac{1}{1+e^{-z}}
# ]

# Where:

# [
# z=b_0+b_1x
# ]

# * (P(y=1)) = Probability of class 1
# * (b_0) = Intercept
# * (b_1) = Coefficient
# * (x) = Input feature

# ---

# ## How Classification Works

# If Probability ≥ 0.5
# → Predict Class 1

# If Probability < 0.5
# → Predict Class 0

# Example:

# | Probability | Prediction |
# | ----------- | ---------- |
# | 0.90        | Pass (1)   |
# | 0.75        | Pass (1)   |
# | 0.30        | Fail (0)   |
# | 0.10        | Fail (0)   |

# ---

# ## Types of Logistic Regression

# ### 1. Binary Logistic Regression

# Only two classes.

# Examples:

# * Pass / Fail
# * Yes / No
# * Spam / Not Spam

# ### 2. Multinomial Logistic Regression

# More than two classes.

# Examples:

# * Red
# * Blue
# * Green

# ### 3. Ordinal Logistic Regression

# Classes have an order.

# Examples:

# * Low
# * Medium
# * High

# ---

# ## Important Terms

# ### Feature (X)

# Input variable used for prediction.

# Example:

# * Study Hours
# * Age
# * Salary

# ### Target (Y)

# Output category.

# Example:

# * Pass/Fail
# * Spam/Not Spam

# ### Probability

# Chance of belonging to a class.

# Example:

# * 0.85 = 85% chance of passing.

# ### Decision Boundary

# The threshold that separates classes.

# Usually:

# Probability = 0.5

# ---

# ## Advantages

# * Easy to implement.
# * Fast training.
# * Works well for binary classification.
# * Outputs probabilities.
# * Easy to interpret.

# ---

# ## Limitations

# * Assumes a linear relationship between features and log-odds.
# * Struggles with highly complex data.
# * Sensitive to outliers.

# ---

# ## Scikit-Learn Implementation

# ```python
# from sklearn.linear_model import LogisticRegression
# d={
#     'marks':[35, 45, 55, 65, 75],
#     'result':[0, 0, 0, 1, 1]
# }
# df=pd.DataFrame(d)

# X = df[["marks"]]
# y = df["result"]

# model = LogisticRegression()
# model.fit(X, y)

# prediction = model.predict([[4]])
# print(prediction)
# ```

# ---

# ## Evaluation Metrics

# ### Accuracy

# Percentage of correct predictions.

# ### Precision

# Out of predicted positives, how many are actually positive.

# ### Recall

# Out of actual positives, how many were correctly identified.

# ### F1 Score

# Harmonic mean of Precision and Recall.

# ### Confusion Matrix

# | Actual \ Predicted | 0  | 1  |
# | ------------------ | -- | -- |
# | 0                  | TN | FP |
# | 1                  | FN | TP |

# Where:

# * TP = True Positive
# * TN = True Negative
# * FP = False Positive
# * FN = False Negative

# ---

# ## Real-World Applications

# * Email Spam Detection
# * Disease Prediction
# * Loan Approval
# * Customer Churn Prediction
# * Fraud Detection
# * Sentiment Analysis

# ---

# ## Conclusion

# Logistic Regression is one of the most important classification algorithms in Machine Learning. It predicts probabilities using the Sigmoid Function and is commonly used for binary classification tasks such as spam detection, disease diagnosis, fraud detection, and pass/fail prediction.
````
