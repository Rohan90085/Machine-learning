```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

le=LabelEncoder()
he=OneHotEncoder()
ss=StandardScaler()
mms=MinMaxScaler()
d={
    'sal':[100,200,300,None,500,600],
    'sex':['male','female','female','male','female','male'],
    'age':[20,30,None,50,60,67],
    'car':['bmw','audi','penz','bmw','audi','bmw']
}
df=pd.DataFrame(d)
df[['sal','age']]=df[['sal','age']].fillna(df[['sal','age']].mean()) #fills multiple null values with mean
# print(df)

df['sex']=le.fit_transform(df['sex'])#labelencoder
df['car']=he.fit_transform(df[['car']]).toarray()#onehotencoder
df['sal']=ss.fit_transform(df[['sal']])# be carefull for all scikitlearn need to give the [[]],mean=0,standarddeviation=1
df['age']=mms.fit_transform(df[['age']])#minmaxscaler
print(df)
student={
    'hours':[1,2,3,4,5,],
    'marks':[10,20,30,40,50]
}
sdf=pd.DataFrame(student)
xtrain,xtest,ytrain,ytest=train_test_split(sdf['hours'],sdf['marks'],test_size=0.2,random_state=42)
print("xtrain:\n",xtrain)#gives the selecetd value for the model training
print("ytrain:\n",ytrain)#gives the model selected ouptut to train
print("\ninput taken to test:",xtest)#gives the input taken for test model
print("\noutput of taken value to test:",ytest)#output
print(sdf)#for checking data

```
