```python
import pandas as pd
from sklearn.cluster import KMeans
```

```python
data={
    'age':[20,30,40,22,38,25],
    'spending':[100,200,300,110,290,130]
}
df=pd.DataFrame(data)
```

```python

kmeans=KMeans(n_clusters=2,random_state=42)
kmeans.fit(df)
print(kmeans.labels_)
print(kmeans.cluster_centers_)

```
